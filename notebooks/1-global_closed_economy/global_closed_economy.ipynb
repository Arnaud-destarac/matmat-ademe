**Exercise:** Verify the system equilibrium

```python
# Check the equation x = Z.1n + Y.1m
# ----------------------------------
# Calculate the columns sum of Z
# help: use the pandas method sum(axis=1), applicable on dataframes
z_sum_on_columns = system.Z.df.sum(axis=1)
# Calculate the columns sum of Y
y_sum_on_columns = system.Y.df.sum(axis=1)
# Check the equation x - (Z + Y) < 0.001
x = system.x.df.squeeze()
x - z_sum_on_columns - y_sum_on_columns < 0.001
```

**Exercise:** Verify the Leontief equation

```python
# Check the equation x = L.Y
# --------------------------
# Calculate L.Y
L = system.L.df
Y = system.Y.get_domestic_origin()
# help: you can use @ to perform a matricial product
production = L @ Y
# Check the equation x = L.Y
# i.e. check that x - L.Y < 0.001
x = system.x.df.squeeze()
difference = x - production.sum(axis=1)
# Check difference
tolerance = 0.001
print(f"The difference is less than {tolerance}: {difference < 0.001}")
```
