**Exercise:** Verify the Leontief equation

```python
# Check the equation x = L_k.Y_wo_GFCF
# --------------------------
# Calculate L_k.Y_wo_GFCF
L_k = system.L_k.get_domestic_origin(keep_origin=True)
Y_wo_gfcf = system.Y.drop_gfcf(inplace=False).get_domestic_origin()
production = L_k @ Y_wo_gfcf
# Check the equation x = L_k.Y_wo_GFCF
# i.e. check that x - L_k.Y_wo_GFCF < 0.001
x = system.x.get_domestic_origin().squeeze()
difference = x - production.sum(axis=1)
# Check difference
tolerance = 0.001
print(f"The difference is less than {tolerance}:\n{difference < 0.001}")
```
