```python
import pandas as pd
```

## Load detail levels

```python
from matmat.core.detail_level import core as dl, factory as dl_factory
```

```python
dls = dl_factory.make_dl_dict_from_path(path="../../data/4-PlaneFR/1-accounts/Exiobase/3.10.2_pxp_2019", file_name="detail_levels.xlsx")
```

```python
regions = dls["regions"]
sectors = dls["sectors"]
final_demand_categories = dls["final_demand_categories"]
```

## Build a system

```python
from matmat.core.accounts.system import builder as system_builder
system_director = system_builder.get_director(reset=True)
```

```python
system_director.set_regions(regions)
system_director.set_sectors(sectors)
system_director.set_final_demand_categories(final_demand_categories)
```

```python
system = system_director.make_exo_invest_matrix_system()
```

```python
system.id.base_year = 2019
```

## Build an extension

```python
from matmat.core.accounts.extension import builder as extension_builder
extension_director = extension_builder.get_director(reset=True)
```

```python
extension_director.set_regions(regions)
extension_director.set_sectors(sectors)
extension_director.set_final_demand_categories(final_demand_categories)
```

### Build an extension categories detail level

```python
ghg_emissions_df = pd.DataFrame({"perimeter": ["combustion", "other"]})
```

```python
ghg_emissions_df
```

```python
ghg_emissions_dl = dl.ExtensionCategoriesDL(extension_name="ghg_emissions", df=ghg_emissions_df)
```

```python
ghg_emissions_dl.df
```

### Use the detail level to build an extension

```python
extension_director.set_extension_categories(ghg_emissions_dl)
```

```python
ghg_emissions = extension_director.make_use_based_extension(name="ghg_emissions")
```

```python
ghg_emissions.F_Y.df
```

## Build an accounts from a system and an extension

```python
from matmat.core.accounts import builder as accounts_builder
accounts_director = accounts_builder.get_director(reset=True)
```

```python
accounts = accounts_director.make_from_system_and_extensions(
    system=system,
    extensions={water.name: water}
)
```
